```python
#  Predicting the grade
score=int(input())
if score>=90:
  grade='A'
elif score>=80:
  grade='B'
elif score>=70:
  grade='C'
elif score>=60:
  grade='D'
else:
  grade='F'
print(grade)
```

```python
# Checking even or odd and checking wherther divisble by both 3 and 5 or only 3 or only 5

n=int(input("Enter a number"))
if (n%2==0):
  print("Even")
else:
  print("Odd")
if (n%3==0 and n%5==0):
  print('Divisible by by 3 and 5')
else:
  if (n%3==0):
    print('Only divisible by 3')
  else:
    print('Only divisible by 5')
```
